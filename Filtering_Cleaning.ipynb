```python
# colums we plsn to use for the football project

football_columns = [
    "Div", "Date", "Time",
    "HomeTeam", "AwayTeam",
    "FTHG", "FTAG", "FTR",
    "HTHG", "HTAG", "HTR",
    "Referee",
    "HS", "AS",
    "HST", "AST",
    "HF", "AF",
    "HC", "AC",
    "HY", "AY",
    "HR", "AR"
]

# chek misssing values onaly in the colums we plan to use 

print(df[football_columns].isnull().sum())
```

```python
# kepp only the colums in football_columns 

df =  df[football_columns].copy()

# check the new size of the dataset
print(df.shape)

df.head()
```

```python
# check the data type

df.dtypes
```

```python
#Convet Date form str to int64

df["Date"] = pd.to_datetime(df["Date"], format= "%d/%m/%Y", errors= "coerce")

# check Date type
print(df["Date"].dtypes)
```

```python
# check if all Date were converted
print("invalid dates: ", df["Date"].isnull().sum())
```

```python
# check uniques match result value

print("Full-time result: ", df["FTR"].unique())
print("Half-time result: ", df["HTR"].unique())
```

```python
# check for negarive values in numeir football columns

numeric_colums = [
    "FTHG", "FTAG",
    "HTHG", "HTAG",
    "HS", "AS",
    "HST", "AST",
    "HF", "AF",
    "HC", "AC",
    "HY", "AY",
    "HR", "AR"
]


for c in numeric_colums:
    print(c, (df[c] < 0).sum())
```

```python
# check if the shots on target HST are greter than the total shots HS

home_shot_errors = (df["HST"] > df["HS"]).sum()
away_shot_errors = (df["AST"] > df["AS"]).sum()

print("Home shot errors:", home_shot_errors)
print("Away shot errors:", away_shot_errors)


```

```python
# chack the distrbution of match result

print(df["FTR"].value_counts())

print(df["FTR"].value_counts(normalize=True) * 100)
```

```python
# check the all the unique team names

team = sorted(set(df["HomeTeam"]) | set(df["AwayTeam"]))

print("Number of team: ", len(team))
print(team)
```

```python
#check if FTR matches the fulltime score

result_errors = df[

    ((df["FTHG"] > df["FTAG"]) & (df["FTR"] != "H")) |
    ((df["FTHG"] < df["FTAG"]) & (df["FTR"] != "A")) |
    ((df["FTHG"] == df["FTAG"]) & (df["FTR"] != "D"))
]

print("Number of result errors: ", len(result_errors))

result_errors

```

```python
#check if HTR matches the halftime score

half_time_errors = df[
    ((df["HTHG"] > df["HTAG"]) & (df["HTR"] != "H")) |
    ((df["HTHG"] < df["HTAG"]) & (df["HTR"] != "A")) |
    ((df["HTHG"] == df["HTAG"]) & (df["HTR"] != "D"))
]

print("Number of half-time result errors:", len(half_time_errors))

half_time_errors
```

```python
#final check for duplicates and missing values

print("Duplicate rows:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
print("Final dataset shape:", df.shape)
```
